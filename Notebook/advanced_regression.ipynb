```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_1=pd.read_csv("../Data/train.csv")
df_2=pd.read_csv("../Data/test.csv")
if 'SalePrice' not in df_2.columns:
  df_2['SalePrice']=0
df=pd.concat([df_1,df_2],axis=0)
df=df.set_index('Id')
```

```python
df_null=df[df.isnull().sum()[df.isnull().sum()>0].index]
```

```python
object_cols = df.select_dtypes(include=['object']).columns
df_objects = df[object_cols].copy()
high_null_object_cols = df_objects.columns[df_objects.isna().sum() > 1100]
df = df.drop(columns=high_null_object_cols)
df_objects = df_objects.drop(columns=high_null_object_cols)
```

```python
df_objects = df_objects.fillna('null')
df_objects_encoded = pd.get_dummies(df_objects)
```

```python
df_objects_encoded.head()
```

```python
for i in df_objects_encoded.columns:
    if 'null' in i:
        df_objects_encoded=df_objects_encoded.drop(i,axis=1)
        print(i)
```

```python
df_numeric = df.select_dtypes(exclude=['object']).copy()
new_df = pd.concat([df_numeric, df_objects_encoded], axis=1)
```

```python
len(new_df.columns)
```

```python
len(df.columns),len(df_objects_encoded.columns)
```

```python
missing_counts = new_df.isna().sum()
missing_counts[missing_counts > 0]
```

```python
Mode_columns = ['GarageCars', 'GarageYrBlt', 'BsmtFullBath', 'BsmtHalfBath']
Mean_columns = ['LotFrontage', 'MasVnrArea', 'BsmtFinSF1', 'BsmtFinSF2',
                'BsmtUnfSF', 'TotalBsmtSF', 'GarageArea']

for col in Mode_columns:
    if col in new_df.columns:
        new_df[col] = new_df[col].fillna(new_df[col].mode()[0])
    else:
        print(f"⚠️ Column not found, skipping: {col}")

for col in Mean_columns:
    if col in new_df.columns:
        new_df[col] = new_df[col].fillna(round(new_df[col].mean(), 2))
    else:
        print(f"⚠️ Column not found, skipping: {col}")

print(new_df.isna().sum()[new_df.isna().sum() > 0])
```

```python
training_data=new_df[0:len(df_1)]
testing_data=new_df[len(df_1):]
testing_data=testing_data.drop(columns='SalePrice')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error 
from sklearn.linear_model import LinearRegression
from xgboost import XGBRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
X = training_data.drop(columns='SalePrice')
y = training_data['SalePrice']
X_train, X_test, Y_train, Y_test = train_test_split(X, y, test_size = 0.2)
Y_train = np.reshape(Y_train,(-1, 1))
Y_test = np.reshape(Y_test,(-1, 1))
X_train.shape, Y_train.shape
```

```python

model_1 = LinearRegression()
model_1.fit(X, y)
y_pred = model_1.predict(X_test)
mean_squared_error(Y_test, y_pred)
```

```python
model_2 = XGBRegressor(n_estimators=1000, learning_rate=0.1, random_state=42)
model_2.fit(X, y)
y_pred = model_2.predict(X_test)
mean_squared_error(Y_test, y_pred)
```

```python
model_3 = RandomForestRegressor(n_estimators=1000)
model_3.fit(X, y)
y_pred = model_3.predict(X_test)
mean_squared_error(Y_test, y_pred)
```

```python
sort_indices = np.argsort(Y_test)

# Plot
plt.figure(figsize=(10, 6))
plt.plot(np.arange(len(Y_test)), Y_test, label='Original')
plt.plot(np.arange(len(Y_test)), y_pred, label='Predicted')
plt.xlabel('Index')
plt.ylabel('Price')
plt.title('Original vs Predicted Prices')
plt.legend()
plt.show()
```
